## 1. Load Raw Datasets
Inspect files in `data/raw/` to understand tables and schemas.

```python
import pandas as pd, os, glob
raw_files = glob.glob('../data/raw/*.csv')
for f in sorted(raw_files):
    df = pd.read_csv(f)
    print(f'{os.path.basename(f)}: {df.shape[0]} rows, {df.shape[1]} cols')
```

### Observation & Business Implications
- **Observation**: The dataset contains 9 relational tables with ~100,000 orders placed across Brazil between 2016 and 2018.
- **Insight**: Multi-table structure requires joining on `order_id`, `customer_id`, `product_id`, and `seller_id`.
