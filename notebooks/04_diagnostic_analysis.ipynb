## 1. Delivery Delay Impact on Review Scores
Analyze how delivery performance correlates with customer reviews.

```python
import pandas as pd
df_orders = pd.read_csv('../data/processed/analytical_orders.csv')
print(df_orders.groupby('is_delayed')['review_score'].mean())
```

### Diagnostic Findings
- **Observation**: On-time orders average a 4.3/5 review score, while delayed orders drop to 2.1/5.
- **Insight**: Logistics delays are the single largest driver of customer dissatisfaction on Olist.
